```python
#Daftar library yang digunakan

# 1. Library bawaan Python untuk matematika dan simulasi dasar
import math
# Menghitung banyak cara memilih 2 server dari total 5 server
kombinasi = math.comb(5, 2)
print(f"1. math  : Banyak kombinasi memilih 2 dari 5 server adalah {kombinasi} cara.")
```

```python
import random
# 2. Menentukan status server acak dengan peluang uptime 90% (0.9)
angka_acak = random.random() # Menghasilkan angka antara 0.0 - 1.0
status_server = "UP" if angka_acak <= 0.9 else "DOWN"
print(f"2. random: Status simulasi server saat ini adalah {status_server}.")
```

```python
# 3.. Library untuk komputasi numerik dan pemrosesan array
import numpy as np
# Menghasilkan 5 sampel acak waktu respons API (rata-rata 120ms, standar deviasi 15ms)
sampel_waktu = np.random.normal(loc=120, scale=15, size=5)
print(f"3. numpy : 5 sampel acak waktu respons (ms): {np.round(sampel_waktu, 2)}")
```

```python
# 4. Library untuk perhitungan probabilitas dan statistika (berbagai distribusi)
from scipy.stats import binom, expon, norm, poisson
# Menghitung probabilitas 2 sensor diretas dari total 5 sensor (peluang diretas 10%)
prob_binom = binom.pmf(k=2, n=5, p=0.1)
print(f"4. scipy : Probabilitas tepat 2 sensor diretas (Binomial): {prob_binom:.4f}")
```

```python
# 5. Library untuk visualisasi data (grafik)
import matplotlib.pyplot as plt
#Menggunakan seaborn agar tampilan grafik lebih rapi/estetik
import seaborn as sns
sns.set_theme(style="whitegrid")
print("Semua library berhasil diimpor dan siap digunakan!")
# Membuat visualisasi Kurva Normal untuk waktu respons
print("5. matplotlib: Menampilkan grafik kurva normal...")

x_axis = np.linspace(80, 160, 100) # Rentang nilai sumbu X (waktu)
y_axis = norm.pdf(x_axis, loc=120, scale=15) # Menghitung nilai sumbu Y (kepadatan)

plt.figure(figsize=(6, 3))
plt.plot(x_axis, y_axis, color='blue', lw=2)
plt.title('Contoh Penggunaan Matplotlib (Kurva Normal)')
plt.xlabel('Waktu Respons (ms)')
plt.ylabel('Kepadatan Probabilitas (PDF)')
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```

```python
# Probabilitas prior
p_spam = 0.20
p_normal = 0.80

# Probabilitas bersyarat (Likelihood)
p_gratis_spam = 0.80
p_gratis_normal = 0.05

# Menghitung Probabilitas Total email mengandung kata "Gratis" (Evidence)
p_gratis_total = (p_gratis_spam * p_spam) + (p_gratis_normal * p_normal)

# Teorema Bayes: Menghitung P(Spam | Gratis) (Posterior)
p_spam_jika_gratis = (p_gratis_spam * p_spam) / p_gratis_total

print(f"Probabilitas email adalah spam jika terdapat kata 'Gratis': {p_spam_jika_gratis:.4f} ({p_spam_jika_gratis*100:.2f}%)")
```

```python
total_server = 10
server_down = 3
server_dipilih = 2

# Kombinasi memilih 2 server dari 3 server yang down
kombinasi_keduanya_down = math.comb(server_down, server_dipilih)

# Kombinasi total memilih 2 server dari keseluruhan 10 server
kombinasi_total = math.comb(total_server, server_dipilih)

# Menghitung peluang probabilitas klasik
prob_keduanya_down = kombinasi_keduanya_down / kombinasi_total

print(f"Banyak cara memilih 2 dari {server_down} server down: {kombinasi_keduanya_down}")
print(f"Total cara memilih 2 dari {total_server} server: {kombinasi_total}")
print(f"Probabilitas 2 server terpilih keduanya down: {prob_keduanya_down:.4f} ({prob_keduanya_down*100:.2f}%)")
```

```python
jumlah_simulasi = 10000
hari_aman = 0
prob_uptime = 0.9

# 1. Pendekatan Simulasi Acak (Monte Carlo)
for _ in range(jumlah_simulasi):
    # Bernilai True jika angka acak (0.0 - 1.0) <= 0.9
    server1 = random.random() <= prob_uptime
    server2 = random.random() <= prob_uptime
    server3 = random.random() <= prob_uptime

    # Aplikasi aman jika jumlah server UP (True) >= 2
    if sum([server1, server2, server3]) >= 2:
        hari_aman += 1

prob_simulasi = hari_aman / jumlah_simulasi

# 2. Pendekatan Matematis Teoritis (Distribusi Binomial)
# Peluang aman = Peluang tepat 2 server UP + Peluang tepat 3 server UP
prob_matematis = binom.pmf(2, 3, prob_uptime) + binom.pmf(3, 3, prob_uptime)

print(f"Probabilitas sistem aman (Simulasi {jumlah_simulasi} iterasi): {prob_simulasi:.4f}")
print(f"Probabilitas sistem aman (Perhitungan Matematis): {prob_matematis:.4f}")
```

```python
# Parameter Distribusi Binomial
n_nodes = 20      # Total percobaan (jumlah sensor)
p_attack = 0.1    # Probabilitas 'sukses' (sensor diretas)
k_target = 3      # Target kejadian (tepat 3 sensor)

# Menghitung Probabilitas spesifik P(X=3) menggunakan PMF (Probability Mass Function)
prob_tepat_3 = binom.pmf(k_target, n_nodes, p_attack)

# Menghitung Nilai Harapan (Mean) dan Varians (Parameter 'mv' = mean, variance)
nilai_harapan, varians = binom.stats(n_nodes, p_attack, moments='mv')

print(f"Probabilitas tepat {k_target} node diserang : {prob_tepat_3:.4f} ({prob_tepat_3*100:.2f}%)")
print(f"Nilai Harapan (Ekspektasi)         : {nilai_harapan} node")
print(f"Varians Distribusi                 : {varians}")

# Visualisasi Distribusi Probabilitas
x_kejadian = np.arange(0, 11) # Menampilkan rentang X (0 hingga 10) untuk kejelasan grafik
y_probabilitas = binom.pmf(x_kejadian, n_nodes, p_attack)

plt.figure(figsize=(8, 4))
plt.bar(x_kejadian, y_probabilitas, color='coral', edgecolor='black', alpha=0.8)
plt.axvline(nilai_harapan, color='red', linestyle='dashed', linewidth=2, label=f'Nilai Harapan = {nilai_harapan}')
plt.title('Distribusi Binomial: Ancaman Peretasan Sensor IoT')
plt.xlabel('Jumlah Sensor IoT yang Diretas')
plt.ylabel('Probabilitas (PMF)')
plt.xticks(x_kejadian)
plt.legend()
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.show()
```

```python
# Parameter Distribusi Normal
mean_api = 120    # Rata-rata waktu respons (mu)
std_api = 15      # Standar deviasi (sigma)
batas_alert = 150 # Ambang batas kritis

# Menghitung probabilitas memicu alert (P(X > 150))
# Menggunakan Survival Function (sf) yang setara dengan 1 - CDF
prob_alert = norm.sf(batas_alert, loc=mean_api, scale=std_api)

print(f"Probabilitas memicu alert (>150ms): {prob_alert:.4f} ({prob_alert*100:.2f}%)")

# Visualisasi Distribusi Normal dan Area Kritis
# Membuat rentang nilai sumbu X dari (mean - 4*sigma) hingga (mean + 4*sigma)
x_norm = np.linspace(mean_api - 4*std_api, mean_api + 4*std_api, 1000)
y_norm = norm.pdf(x_norm, loc=mean_api, scale=std_api)

plt.figure(figsize=(8, 4))
plt.plot(x_norm, y_norm, color='purple', lw=2, label='Kurva PDF Waktu Respons')

# Mengarsir area kritis yang memicu alert (X > 150)
x_kritis = np.linspace(batas_alert, mean_api + 4*std_api, 100)
y_kritis = norm.pdf(x_kritis, loc=mean_api, scale=std_api)
plt.fill_between(x_kritis, y_kritis, color='red', alpha=0.5, label='Area Kritis / Alert (> 150ms)')

# Garis vertikal untuk nilai rata-rata
plt.axvline(mean_api, color='black', linestyle='--', label=f'Rata-rata = {mean_api}ms')

plt.title('Distribusi Normal: Waktu Respons API')
plt.xlabel('Waktu Respons (ms)')
plt.ylabel('Kepadatan Probabilitas (Densitas)')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```
