# Uji Validitas dan Reliabilitas Instrumen
Mata Kuliah Statistika Probabilitas

```python
import re
import csv
import pandas as pd
import numpy as np

from scipy import stats
from google.colab import files

print("Persiapan pustaka berhasil!")
```

```python
up = files.upload()

nama_file = list(up.keys())[0]

print("File terbaca:", nama_file)
```

```python
# Membaca file Excel
df_raw = pd.read_excel(nama_file, header=None)

# Memperbaiki format jika seluruh data terbaca dalam satu kolom
if df_raw.shape[1] == 1:

    isi = df_raw.iloc[:, 0].dropna().astype(str).tolist()

    baris = list(csv.reader(isi))

    df = pd.DataFrame(baris[1:], columns=baris[0])

else:
    df = pd.read_excel(nama_file)

# Membersihkan spasi pada nama kolom
df.columns = df.columns.astype(str).str.strip()

# Mengambil kolom yang berisi butir pernyataan
kolom_butir = [
    kolom for kolom in df.columns
    if "[" in kolom and "]" in kolom
]

butir = df[kolom_butir].copy()

print("Jumlah responden :", len(df))
print("Jumlah butir     :", len(butir.columns))
print("Data kosong      :", int(butir.isna().sum().sum()))

print("\nDaftar kolom butir:")
for i, kolom in enumerate(kolom_butir, 1):
    print(f"{i}. {kolom}")
```

```python
skala = {
    "Sangat Tidak Setuju (STS)": 1,
    "Tidak Setuju (TS)": 2,
    "Netral (N)": 3,
    "Setuju (S)": 4,
    "Sangat Setuju (SS)": 5
}

num = butir.apply(lambda s: s.map(skala))

print(
    "Sel yang gagal dikonversi:",
    int(num.isna().sum().sum())
)

display(num.head())
```

```python
dimensi = {}

for kolom in butir.columns:

    nama = re.match(
        r"^(.*?)\s*\[",
        kolom
    ).group(1).strip().rstrip(".").strip()

    # Menyamakan penulisan nama dimensi
    if nama == "Rsspon emosional":
        nama = "Respon emosional"

    dimensi.setdefault(nama, []).append(kolom)


print("HASIL PENGELOMPOKAN DIMENSI\n")

for nama, daftar in dimensi.items():

    print(f"{nama}: {len(daftar)} butir")

print("\nJumlah dimensi:", len(dimensi))
print("Jumlah seluruh butir:", sum(len(v) for v in dimensi.values()))
```

```python
# Menghitung r tabel
n = len(num)

t = stats.t.ppf(0.975, n - 2)

r_tabel = t / np.sqrt((n - 2) + t**2)

print(f"Jumlah responden = {n}")
print(f"r tabel (α=0,05; df={n-2}) = {r_tabel:.3f}")


# Pengujian validitas
hasil_validitas = []

for nama, daftar in dimensi.items():

    x = num[daftar]

    total = x.sum(axis=1)

    for i, kolom in enumerate(daftar, 1):

        # Korelasi Pearson item-total
        r, p = stats.pearsonr(
            x[kolom],
            total
        )

        # Corrected item-total correlation
        r_koreksi, _ = stats.pearsonr(
            x[kolom],
            total - x[kolom]
        )

        hasil_validitas.append({

            "Dimensi": nama,
            "Butir": f"B{i}",
            "Pernyataan": kolom.split("[", 1)[1].rstrip("]"),
            "r hitung": round(r, 3),
            "Corrected r": round(r_koreksi, 3),
            "p-value": round(p, 4),

            "Keterangan": (
                "Valid" if r > r_tabel
                else "Tidak valid"
            )

        })


tabel_validitas = pd.DataFrame(hasil_validitas)

display(tabel_validitas)
```

```python
# Fungsi menghitung Cronbach's Alpha
def cronbach_alpha(x):

    k = x.shape[1]

    if k < 2:
        return np.nan

    varians_butir = x.var(ddof=1).sum()

    varians_total = x.sum(axis=1).var(ddof=1)

    if varians_total == 0:
        return np.nan

    return (
        k / (k - 1)
        * (1 - varians_butir / varians_total)
    )


# Fungsi kategori reliabilitas
def kategori(a):

    if pd.isna(a):
        return "Tidak dapat dihitung"

    return (
        "Sangat baik" if a >= 0.90
        else "Baik" if a >= 0.80
        else "Reliabel" if a >= 0.70
        else "Tidak reliabel"
    )


hasil_reliabilitas = []
alpha_dihapus = []


for nama, daftar in dimensi.items():

    x = num[daftar]

    a = cronbach_alpha(x)

    hasil_reliabilitas.append({

        "Dimensi": nama,
        "Jumlah butir": len(daftar),
        "Cronbach's alpha": round(a, 3),
        "Kategori": kategori(a)

    })

    # Menghitung alpha jika setiap butir dihapus
    for i, kolom in enumerate(daftar, 1):

        if len(daftar) > 2:

            alpha_baru = cronbach_alpha(
                x.drop(columns=kolom)
            )

            alpha_dihapus.append({

                "Dimensi": nama,
                "Butir": f"B{i}",
                "Alpha jika dihapus": (
                    round(alpha_baru, 3)
                    if pd.notna(alpha_baru)
                    else np.nan
                )

            })


# Membuat tabel hasil
tabel_reliabilitas = pd.DataFrame(
    hasil_reliabilitas
)

tabel_alpha_dihapus = pd.DataFrame(
    alpha_dihapus
)


print("HASIL UJI RELIABILITAS")

display(tabel_reliabilitas)

print("HASIL ALPHA JIKA BUTIR DIHAPUS")

display(tabel_alpha_dihapus)
```

```python
# Menggabungkan hasil validitas dan alpha jika dihapus
gabung = tabel_validitas.merge(
    tabel_alpha_dihapus,
    on=["Dimensi", "Butir"],
    how="left"
)


# Menyimpan hasil ke Excel
nama_output = "hasil_uji_validitas_reliabilitas.xlsx"

with pd.ExcelWriter(nama_output) as writer:

    gabung.to_excel(
        writer,
        sheet_name="Validitas",
        index=False
    )

    tabel_reliabilitas.to_excel(
        writer,
        sheet_name="Reliabilitas",
        index=False
    )


# Mengunduh hasil
files.download(nama_output)

print("Pengujian selesai!")
print("File hasil berhasil dibuat:", nama_output)
```

## Latihan dan Tugas

### Latihan 1 — Interpretasi

```python
# Butir dengan korelasi terendah (kandidat bermasalah)
display(
    tabel_validitas
    .assign(**{"r tabel": round(r_tabel, 3)})
    [["Dimensi", "Butir", "Pernyataan", "r hitung", "r tabel", "Corrected r"]]
    .sort_values("Corrected r")
    .head(5)
)
print("Butir tidak valid (r hitung <= r tabel):",
      int((tabel_validitas["r hitung"] <= r_tabel).sum()))
```

### Latihan 2 — Perbandingan kriteria

```python
# Menghitung jumlah butir valid berdasarkan r hitung
valid_r_hitung = (
    tabel_validitas["r hitung"] > r_tabel
).sum()

# Menghitung jumlah butir valid berdasarkan corrected r
valid_corrected = (
    tabel_validitas["Corrected r"] >= 0.30
).sum()

# Membuat tabel perbandingan
perbandingan = pd.DataFrame({
    "Kriteria": [
        "r hitung > r tabel",
        "Corrected r >= 0,30"
    ],
    "Jumlah butir valid": [
        valid_r_hitung,
        valid_corrected
    ]
})

display(perbandingan)

print(
    "Selisih jumlah butir:",
    abs(valid_r_hitung - valid_corrected)
)
```

```python
# Butir yang lolos r hitung tetapi gagal corrected r >= 0,30
display(
    tabel_validitas[tabel_validitas["Corrected r"] < 0.30]
    [["Dimensi", "Butir", "r hitung", "Corrected r"]]
)
```

### Latihan 3 — Perbaikan instrumen

```python
butir_tidak_valid = tabel_validitas[tabel_validitas["Corrected r"] < 0.30]

if len(butir_tidak_valid) > 0:

    target = butir_tidak_valid.sort_values("Corrected r").iloc[0]
    nama_dimensi = target["Dimensi"]
    nomor_butir = target["Butir"]

    daftar_butir = dimensi[nama_dimensi]

    kolom_dihapus = daftar_butir[
        int(nomor_butir.replace("B", "")) - 1
    ]

    # Data sebelum penghapusan
    data_sebelum = num[daftar_butir]
    alpha_sebelum = cronbach_alpha(data_sebelum)

    # Data sesudah penghapusan
    daftar_baru = [k for k in daftar_butir if k != kolom_dihapus]
    data_sesudah = num[daftar_baru]
    alpha_sesudah = cronbach_alpha(data_sesudah)

    hasil_perbaikan = pd.DataFrame({
        "Keterangan": ["Sebelum penghapusan", "Sesudah penghapusan"],
        "Jumlah butir": [len(daftar_butir), len(daftar_baru)],
        "Cronbach's Alpha": [round(alpha_sebelum, 3), round(alpha_sesudah, 3)]
    })

    print("Dimensi            :", nama_dimensi)
    print("Butir yang dihapus :", nomor_butir)
    print("Pernyataan         :", target["Pernyataan"])
    print("Corrected r        :", target["Corrected r"])

    display(hasil_perbaikan)

else:
    print("Tidak ditemukan butir bermasalah.")
```

### Latihan 4 — Pengaruh ukuran sampel

```python
# Mengambil sampel acak sebanyak 20 responden
num_sample = num.sample(
    n=20,
    random_state=1
)

# Menghitung r tabel sampel
n_sample = len(num_sample)

t_sample = stats.t.ppf(
    0.975,
    n_sample - 2
)

r_tabel_sample = t_sample / np.sqrt(
    (n_sample - 2) + t_sample**2
)

# Menguji validitas ulang
hasil_sample = []

for nama, daftar in dimensi.items():

    x = num_sample[daftar]
    total = x.sum(axis=1)

    for i, kolom in enumerate(daftar, 1):

        r, p = stats.pearsonr(
            x[kolom],
            total
        )

        hasil_sample.append({
            "Dimensi": nama,
            "Butir": f"B{i}",
            "r hitung": round(r, 3),
            "Keterangan": (
                "Valid"
                if r > r_tabel_sample
                else "Tidak valid"
            )
        })

tabel_sample = pd.DataFrame(hasil_sample)

print("Jumlah sampel:", n_sample)
print("r tabel sampel:", round(r_tabel_sample, 3))

display(tabel_sample)

print(
    "Jumlah butir valid:",
    (tabel_sample["Keterangan"] == "Valid").sum()
)

print(
    "Jumlah butir tidak valid:",
    (tabel_sample["Keterangan"] == "Tidak valid").sum()
)

# Perbandingan dengan data penuh
print()
print(f"r tabel      : n={len(num)} -> {r_tabel:.3f} | n={n_sample} -> {r_tabel_sample:.3f}")
print("Butir valid  :", int((tabel_validitas['Keterangan'] == 'Valid').sum()),
      "->", int((tabel_sample['Keterangan'] == 'Valid').sum()))
print("r hitung min : data penuh =", tabel_validitas['r hitung'].min(),
      "| sampel =", tabel_sample['r hitung'].min())
```

### Latihan 5 — Refleksi

**Mengapa instrumen yang reliabel belum tentu valid?**

Reliabilitas hanya menunjukkan **konsistensi** hasil ukur, sedangkan validitas menunjukkan apakah instrumen **mengukur hal yang dimaksud**. Instrumen bisa konsisten tetapi konsisten mengukur hal yang salah atau bias.

**Contoh sederhana:** timbangan yang selalu menunjukkan 2 kg lebih berat dari berat sebenarnya. Hasilnya konsisten (reliabel), tetapi tidak akurat (tidak valid). Pada kuesioner: butir-butir tentang seberapa sering seseorang memegang ponsel bisa menghasilkan α tinggi, tetapi bila dipakai untuk mengukur "prokrastinasi akademik" maka instrumen itu tidak valid walaupun reliabel.

Pada data ini, α setiap dimensi (0,713-0,817) hanya menunjukkan butir-butir saling berkorelasi. Itu belum membuktikan bahwa butir benar-benar mengukur konstruknya; misalnya butir 17-19 pada dimensi "Kebiasaan penggunaan teknologi digital" isinya soal kepercayaan terhadap AI, sehingga perlu ditinjau validitas isinya.
